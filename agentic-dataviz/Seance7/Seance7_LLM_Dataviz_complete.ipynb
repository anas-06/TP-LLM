## 1.1 — Architecture d'un LLM

### Qu'est-ce qu'un LLM ?

Un **Large Language Model (LLM)** est un réseau de neurones entraîné sur d'immenses corpus de texte pour prédire le prochain token (unité de texte) dans une séquence. Cette capacité de prédiction donne l'illusion de "compréhension" et permet de générer du texte cohérent.

### Concepts clés

#### Tokens
Un **token** est l'unité de base que le modèle traite. Il ne correspond pas exactement à un mot :
- `"visualisation"` → 1 token
- `"job_postings"` → 2–3 tokens
- `"4512"` → 1–4 tokens selon le modèle

En moyenne, 1 token ≈ 0,75 mot en anglais, et légèrement moins en français.

#### Context Window (fenêtre de contexte)
La **fenêtre de contexte** est la quantité maximale de tokens que le modèle peut traiter en une seule fois (entrée + sortie). Pour `llama3.2`, elle est de **128 000 tokens** (~96 000 mots). Dépasser cette limite tronque le contexte, ce qui peut produire des réponses incohérentes.

#### Température
La **température** contrôle le caractère aléatoire de la génération :
- `température = 0.0` → réponses déterministes, toujours les mêmes
- `température = 0.7` → bon équilibre créativité/cohérence (valeur par défaut)
- `température = 1.5` → très créatif, parfois incohérent

Pour des tâches structurées (JSON, code), on préfère une **température basse (0.0–0.3)**.

#### Modèles locaux vs cloud

| Critère | Local (Ollama) | Cloud (OpenAI, Anthropic) |
|---|---|---|
| **Coût** | Gratuit | Payant (par token) |
| **Confidentialité** | Données restent en local | Données envoyées au serveur |
| **Performance** | Limitée au matériel local | Très haute (GPU cloud) |
| **Débit** | Lent sur CPU | Rapide |
| **Modèles dispo** | llama3.2, mistral, gemma… | GPT-4o, Claude Opus… |

Dans ce cours, on utilise **Ollama** pour des raisons de confidentialité et de coût zéro.

### Flux de traitement d'une requête

```
┌─────────────────────────────────────────────────────────────┐
│                    ARCHITECTURE LLM                         │
│                                                             │
│  ┌──────────┐    ┌───────────┐    ┌─────────┐    ┌───────┐ │
│  │  TEXTE   │───▶│TOKENIZER  │───▶│  LLM    │───▶│TOKENS │ │
│  │ d'entrée │    │(découpage)│    │(réseau  │    │sortie │ │
│  │          │    │           │    │de neur.)│    │       │ │
│  └──────────┘    └───────────┘    └─────────┘    └───┬───┘ │
│                                                       │     │
│                                                       ▼     │
│                                               ┌────────────┐│
│                                               │DÉTOKENIZER ││
│                                               │(reconstr.) ││
│                                               └─────┬──────┘│
│                                                     │       │
│                                                     ▼       │
│                                               ┌──────────┐  │
│                                               │  TEXTE   │  │
│                                               │ de sortie│  │
│                                               └──────────┘  │
└─────────────────────────────────────────────────────────────┘
```

### Pourquoi la précision numérique est un risque majeur

Les LLMs **ne calculent pas** : ils **prédisent** du texte. Lorsqu'on demande `"Combien de lignes contient ce CSV ?"`, le modèle ne compte pas — il génère la réponse la plus probable statistiquement d'après son entraînement. Résultat :

- Il peut **inventer un nombre plausible** (hallucination numérique)
- Il peut **arrondir** de façon erronée
- Il peut **extrapoler** à partir d'exemples vus à l'entraînement

> **⚠️ Règle d'or : ne jamais faire confiance à un chiffre produit par un LLM sans vérification indépendante via pandas/numpy.**

## 1.2 — Prompting structuré pour la data

### Stratégies de prompting

#### Zero-shot
On donne l'instruction directement, sans exemple. Simple mais moins précis pour des tâches complexes.

```
Recommande une visualisation pour ce dataset.
```

#### Few-shot
On donne 1 à 5 exemples du format attendu avant la vraie question. Le modèle comprend mieux le format souhaité.

```
Exemple 1 : Dataset de ventes → Bar chart par région
Exemple 2 : Dataset de températures → Line chart temporel
Maintenant, pour ce dataset d'offres d'emploi :
```

#### Chain-of-thought (CoT)
On demande au modèle de **raisonner étape par étape** avant de répondre. Améliore significativement la qualité sur des tâches analytiques.

```
Réfléchis étape par étape :
1. Quelles sont les variables disponibles ?
2. Quelles questions analytiques sont pertinentes ?
3. Quelle visualisation répond le mieux à chaque question ?
```

### Forçage de sortie JSON structurée

Pour automatiser le traitement des réponses, on peut **forcer** le LLM à produire du JSON en :
1. Définissant le schéma attendu dans le prompt
2. Donnant un exemple de sortie
3. Terminant le prompt par `"Réponds UNIQUEMENT avec du JSON valide, sans explication."`

Exemple de prompt structuré :
```
Tu es un expert en datavisualisation. Analyse ce dataset et réponds
UNIQUEMENT avec un JSON valide ayant cette structure exacte :
{
  "recommandations": [
    {
      "type": "bar chart",
      "colonnes": ["col1", "col2"],
      "justification": "...",
      "priorite": 1
    }
  ]
}
```

### Exemple complet — Prompt structuré avec Ollama

La cellule suivante est une **démonstration complète** : chargement des données, envoi au LLM, parsing de la réponse JSON.

````python
# NE PAS MODIFIER - Démonstration
# Exemple complet : prompt structuré pour obtenir une réponse JSON

import ollama
import pandas as pd
import json
import re

# --- Chargement d'un petit échantillon du dataset ---
df = pd.read_csv('job_postings.csv')
sample = df.head(10)

# Résumé des colonnes pour le prompt (on n'envoie pas toutes les données brutes)
colonnes_info = ", ".join(df.columns.tolist())
apercu = sample.to_string(index=False, max_cols=8)

# --- Construction du prompt structuré ---
prompt = f"""Tu es un expert en datavisualisation. Voici un aperçu d'un dataset d'offres d'emploi.

Colonnes disponibles : {colonnes_info}

Aperçu des 10 premières lignes :
{apercu}

Analyse ce dataset et réponds UNIQUEMENT avec un JSON valide, sans aucune explication ni texte avant ou après.
Le JSON doit avoir cette structure exacte :
{{
  "analyse": {{
    "type_dataset": "description en une phrase",
    "variable_cle": "la colonne la plus intéressante pour analyser",
    "recommandation_principale": "type de visualisation recommandé",
    "justification": "explication en 2-3 phrases"
  }}
}}"""

# --- Appel à l'API Ollama ---
print("Envoi de la requête à Ollama (llama3.2)...")
response = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': prompt}]
)

# --- Récupération du contenu texte ---
contenu_brut = response['message']['content']
print("Réponse brute du LLM :")
print(contenu_brut)
print("-" * 50)

# --- Parsing JSON robuste (le LLM peut envelopper le JSON dans du texte ou du markdown) ---
def extraire_json(texte):
    """Extrait et parse un bloc JSON depuis une réponse LLM potentiellement bruitée."""
    # Étape 1 : supprimer les balises markdown ```json ... ```
    texte_clean = re.sub(r"```(?:json)?\s*", "", texte).replace("```", "").strip()

    # Étape 2 : essai de parsing direct
    try:
        return json.loads(texte_clean)
    except json.JSONDecodeError:
        pass

    # Étape 3 : extraire le premier bloc {...} complet (gestion des accolades imbriquées)
    profondeur, debut = 0, -1
    for idx, car in enumerate(texte_clean):
        if car == "{":
            if profondeur == 0:
                debut = idx
            profondeur += 1
        elif car == "}":
            profondeur -= 1
            if profondeur == 0 and debut != -1:
                candidat = texte_clean[debut:idx + 1]
                try:
                    return json.loads(candidat)
                except json.JSONDecodeError:
                    pass  # bloc trouvé mais encore invalide
                break

    return None  # échec total

resultat = extraire_json(contenu_brut)
if resultat is None:
    print("⚠️  Impossible de parser le JSON — réponse brute affichée ci-dessus.")

# --- Affichage structuré ---
if resultat:
    analyse = resultat.get('analyse', {})
    print(f"Type de dataset    : {analyse.get('type_dataset', 'N/A')}")
    print(f"Variable clé       : {analyse.get('variable_cle', 'N/A')}")
    print(f"Recommandation     : {analyse.get('recommandation_principale', 'N/A')}")
    print(f"Justification      : {analyse.get('justification', 'N/A')}")
````

## 1.3 — Limites et biais des LLMs en analyse de données

### Hallucination numérique

Le phénomène d'**hallucination** désigne le fait qu'un LLM génère des informations fausses mais présentées avec confiance. Pour les données numériques, c'est particulièrement dangereux :

- **Comptage inventé** : `"Il y a environ 30 000 lignes"` alors qu'il y en a 25 114
- **Statistiques fabriquées** : `"Le salaire moyen est de 65 000$"` sans avoir calculé quoi que ce soit
- **Tendances imaginées** : `"Le secteur tech représente 45% des offres"` sans agrégation réelle

### Biais de sélection

Quand on envoie seulement un échantillon (`head(10)`) au LLM, il généralise à partir de cet échantillon :
- Si les 10 premières lignes sont toutes du secteur tech, le LLM pourrait conclure que c'est un dataset tech
- Les valeurs manquantes peuvent être interprétées comme une absence de données dans toute la colonne

### Limites de la fenêtre de contexte

Un CSV de 25 114 lignes fait plusieurs **mégaoctets** de texte, bien au-delà de ce qu'on peut envoyer efficacement :
- Envoyer 1 000 lignes ≈ 50 000–80 000 tokens (selon les données)
- Le coût de traitement augmente quadratiquement avec la taille du contexte
- La qualité de la réponse **décroît** si le contexte est trop long

**Bonne pratique :** Envoyer un résumé statistique (`df.describe()`, liste des colonnes, quelques exemples) plutôt que les données brutes.

### Démonstration : l'hallucination numérique en action

La cellule suivante illustre ce qui se passe quand on demande au LLM de "compter" quelque chose sans lui donner la vraie valeur.

````python
# NE PAS MODIFIER - Démonstration
# Illustration de l'hallucination numérique

import ollama
import pandas as pd

# Chargement du dataset pour connaître les vraies valeurs
df = pd.read_csv('job_postings.csv')

# --- Question au LLM SANS lui donner le nombre réel ---
prompt_sans_info = """J'ai un fichier CSV appelé job_postings.csv qui contient des offres d'emploi.
Sans que je te donne plus d'information, combien de lignes penses-tu qu'il contient ?
Donne un chiffre précis avec ta meilleure estimation."""

print("=" * 60)
print("QUESTION sans information réelle :")
print("=" * 60)
response_sans_info = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': prompt_sans_info}]
)
print(response_sans_info['message']['content'])

print()
print("=" * 60)
print(f"VÉRITÉ TERRAIN (pandas) : {len(df)} lignes")
print("=" * 60)
print()

# --- Autre exemple : demander une moyenne sans données ---
prompt_moyenne = """Pour un dataset d'offres d'emploi avec des colonnes Minimum Pay et Maximum Pay,
quel est le salaire moyen approximatif en dollars américains ?
Donne un chiffre précis."""

print("DEMANDE de moyenne sans données réelles :")
response_moyenne = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': prompt_moyenne}]
)
print(response_moyenne['message']['content'])

print()
# Calcul réel
salaire_moyen_min = df['Minimum Pay'].dropna().mean()
salaire_moyen_max = df['Maximum Pay'].dropna().mean()
print(f"VÉRITÉ TERRAIN (pandas) :")
print(f"  Salaire minimum moyen : {salaire_moyen_min:,.2f} $")
print(f"  Salaire maximum moyen : {salaire_moyen_max:,.2f} $")
print()
print(">>> CONCLUSION : Le LLM a inventé des chiffres. Toujours vérifier avec pandas !")
````

````python
# ✏️ COMPLÉTÉ — Exercice 1-B : demander la meilleure visualisation au LLM
import ollama
import pandas as pd

df = pd.read_csv('job_postings.csv')

# --- 1. Description des colonnes : type + exemples de valeurs ---
# Le LLM ne voit pas le fichier : tout ce qu'il sait vient de cette description.
description_colonnes = """
- Job Posting ID (identifiant numérique) : ex. 2701524240 — inutile pour la visualisation
- Job Posting Date (date AAAA-MM-JJ) : de 2017-01-01 à 2021-12-31
- Job Title (catégoriel) : ex. 'Software Engineer', 'Data Engineer', 'Data Scientist'
- Job Title Full (texte) : intitulé complet, ex. 'Senior Azure Data Engineer'
- Job Title Additional Info (texte, souvent vide) : ex. 'Remote', 'Telecommute'
- Job Position Type (catégoriel) : 'Full-time', 'Contract', 'Internship', 'Part-time'...
- Job Position Level (catégoriel ordinal) : 'Internship' < 'Entry level' < 'Associate' < 'Mid-Senior level' < 'Director' < 'Executive'
- Years of Experience (numérique entier) : de 0 à 25 ans
- Job Skills (texte, liste séparée par des virgules) : ex. 'python, sql, aws, agile'
- Job Location (catégoriel géographique) : ex. 'United States', 'Phoenix, AZ'
- Minimum Pay / Maximum Pay (numérique, en dollars) : ex. 95000 / 128500
- Pay Rate (catégoriel) : 'yr' (annuel) ou 'hr' (horaire)
- Number of Applicants (numérique) : nombre de candidats
- Company Name (catégoriel, très nombreux) : ex. 'Optum', 'Paycor'
- Company Industry (catégoriel) : ex. 'Internet', 'Computer Software', 'Financial Services'
- Company Size (catégoriel ordinal) : ex. '1,001-5,000 employees', '10,001+ employees'
"""

# Taux de valeurs manquantes calculé par pandas : information importante pour le LLM
# (une colonne vide à 93 % n'est pas une bonne base de visualisation).
manquants = (df.isna().mean() * 100).round(1)
manquants_txt = "\n".join(f"- {col} : {pct} % de valeurs manquantes"
                          for col, pct in manquants.items() if pct > 0)

# --- 2. Construction du prompt ---
mon_prompt = f"""Tu es un expert en datavisualisation.
Voici un dataset d'offres d'emploi tech américaines ({len(df)} lignes, {df.shape[1]} colonnes).

Description des colonnes :
{description_colonnes}
Valeurs manquantes (calculées avec pandas) :
{manquants_txt}

Question : quelle est la MEILLEURE visualisation unique à créer pour ce dataset ?
Réponds en français avec :
1. Le type de graphique
2. Les colonnes utilisées (axe X, axe Y, couleur) — uniquement des colonnes listées ci-dessus
3. La question analytique à laquelle le graphique répond
4. Une justification en 3 phrases maximum (tiens compte des valeurs manquantes)"""

# --- 3. Envoi à Ollama ---
response = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': mon_prompt}],
    options={'temperature': 0.2},  # basse : on veut une réponse stable et argumentée
)

# --- 4. Affichage ---
reponse_1b = response['message']['content']  # conservée pour la comparaison de l'exercice 5-B
print(reponse_1b)

# --- 5. Vérification automatique : les colonnes citées existent-elles ? ---
citees = [c for c in df.columns if c.lower() in reponse_1b.lower()]
print("\nColonnes réelles citées par le LLM :", citees or "aucune")
````

**Analyse de la réponse (1-B)**

- Le fait de décrire **le type de chaque colonne** (catégoriel, ordinal, numérique, date) permet au LLM de proposer un graphique cohérent avec la nature des données (ex. bar chart pour du catégoriel, line chart pour `Job Posting Date`).
- Le contexte sur les **valeurs manquantes** est essentiel : `Minimum Pay` / `Maximum Pay` ne sont renseignés que pour ~7 % des offres (1 812 sur 25 114). Une recommandation centrée sur les salaires serait donc fragile ; la vérification de la cellule ci-dessus permet de voir si le LLM en a tenu compte.
- Point de vigilance : le LLM peut citer des colonnes qui n'existent pas ou reformuler leur nom ; la liste « Colonnes réelles citées » permet de le détecter.

### 2-A — DÉMONSTRATION

````python
# NE PAS MODIFIER - Démonstration
# Obtenir 3 recommandations de visualisations au format JSON et les parser

import ollama
import pandas as pd
import json
import re

df = pd.read_csv('job_postings.csv')

# On prend un échantillon de 20 lignes pour le contexte
sample = df.head(20)
colonnes = df.columns.tolist()
apercu = sample[['Job Title', 'Company Industry', 'Job Location', 'Job Skills',
                  'Minimum Pay', 'Maximum Pay', 'Job Position Level']].head(5).to_string(index=False)

# Prompt avec forçage JSON
prompt_json = f"""Tu es un expert en datavisualisation avec Python et Plotly.
Voici un dataset d'offres d'emploi :

Colonnes : {', '.join(colonnes)}
Nombre de lignes : {len(df)}
Aperçu :
{apercu}

Propose exactement 3 recommandations de visualisations pertinentes pour analyser ce dataset.

Réponds UNIQUEMENT avec un JSON valide, sans aucun texte avant ou après le JSON.
Structure exacte à respecter :
{{
  "recommandations": [
    {{
      "numero": 1,
      "titre": "Titre descriptif de la visualisation",
      "type_chart": "bar chart | line chart | scatter plot | pie chart | heatmap | ...",
      "colonnes_x": "nom de la colonne pour l'axe X",
      "colonnes_y": "nom de la colonne pour l'axe Y",
      "couleur": "colonne optionnelle pour la couleur (ou null)",
      "question_analytique": "Quelle question cette viz répond-elle ?",
      "justification": "Pourquoi cette visualisation est-elle pertinente ?",
      "priorite": "haute | moyenne | basse"
    }}
  ]
}}"""

print("Envoi de la requête à Ollama...")
response = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': prompt_json}]
)

contenu = response['message']['content']

# --- Parsing JSON robuste ---
resultat = None
try:
    # Tentative directe
    resultat = json.loads(contenu)
except json.JSONDecodeError:
    # Extraction du bloc JSON si le LLM a ajouté du texte
    match = re.search(r'\{.*\}', contenu, re.DOTALL)
    if match:
        try:
            resultat = json.loads(match.group())
        except json.JSONDecodeError:
            print("Impossible de parser le JSON. Réponse brute :")
            print(contenu)

# --- Affichage structuré des recommandations ---
if resultat:
    recommandations = resultat.get('recommandations', [])
    print(f"\n{len(recommandations)} recommandations reçues :")
    print("=" * 60)
    for reco in recommandations:
        print(f"\n[{reco.get('numero', '?')}] {reco.get('titre', 'Sans titre')}")
        print(f"    Type       : {reco.get('type_chart', 'N/A')}")
        print(f"    Axe X      : {reco.get('colonnes_x', 'N/A')}")
        print(f"    Axe Y      : {reco.get('colonnes_y', 'N/A')}")
        print(f"    Couleur    : {reco.get('couleur', 'N/A')}")
        print(f"    Question   : {reco.get('question_analytique', 'N/A')}")
        print(f"    Priorité   : {reco.get('priorite', 'N/A')}")
        print(f"    Justif.    : {reco.get('justification', 'N/A')}")
````

**Observations (2-B)**

- Un JSON **valide** n'est pas forcément **correct** : le contrôle automatique vérifie (1) que le nombre de recommandations est bien 5, (2) que la contrainte temps/industrie est respectée, (3) que les colonnes citées existent réellement. Les écarts éventuels sont visibles dans la sortie ci-dessus.

````python
# NE PAS MODIFIER - Démonstration
# Fonction utilitaire pour extraire du code Python d'une réponse LLM

import re

def extract_python_code(text: str) -> str:
    """Extrait un bloc de code Python d'une réponse LLM.
    
    Stratégies dans l'ordre :
    1. Cherche un bloc ```python ... ```
    2. Cherche un bloc ``` ... ```
    3. Retourne le texte brut si aucun bloc trouvé
    """
    # Cherche un bloc ```python ... ```
    match = re.search(r'```python\n(.*?)```', text, re.DOTALL)
    if match:
        return match.group(1).strip()
    
    # Sinon cherche ``` ... ```
    match = re.search(r'```\n(.*?)```', text, re.DOTALL)
    if match:
        return match.group(1).strip()
    
    # Sinon retourne le texte brut
    return text.strip()

# Test de la fonction
texte_test = """
Voici le code que vous demandez :
```python
import plotly.express as px
fig = px.bar(x=[1,2,3], y=[4,5,6])
fig.show()
```
Ce code crée un graphique simple.
"""

code_extrait = extract_python_code(texte_test)
print("Code extrait avec succès :")
print(code_extrait)
````

````python
# ✏️ COMPLÉTÉ — Exercice 3-B : génération d'un box plot par le LLM
import ollama
import pandas as pd
import plotly.express as px
import plotly.io as pio
import re, os

# Ouvre les graphiques dans le navigateur (contourne le problème fig.show() en Jupyter)
pio.renderers.default = 'browser'

df = pd.read_csv('job_postings.csv')

# Fonction utilitaire (déjà vue en 3-A)
def extract_python_code(text):
    match = re.search(r'```python\n(.*?)```', text, re.DOTALL)
    if match:
        return match.group(1).strip()
    match = re.search(r'```\n(.*?)```', text, re.DOTALL)
    if match:
        return match.group(1).strip()
    return text.strip()

# Fonction de réparation des erreurs LLM fréquentes sur Plotly
def repair_plotly_code(code: str) -> str:
    # Erreur fréquente : category_order= au lieu de category_orders= (dict)
    code = re.sub(r'category_order\s*=\s*\[', 'category_orders={', code)
    # Fermer le dict si le LLM a mis une liste
    code = re.sub(
        r'(category_orders=\{)(.*?)(\])',
        lambda m: m.group(1) + m.group(2) + '}',
        code, flags=re.DOTALL
    )
    # Remplacer fig.show() par sauvegarde HTML + ouverture navigateur
    code = code.replace('fig.show()', "fig.write_html('chart_ex3b.html')\nimport webbrowser; webbrowser.open('chart_ex3b.html')")
    return code

# --- 1. Prompt précis, construit à partir des vraies valeurs ---
# Ordre logique des niveaux (du plus junior au plus senior) : sans lui, Plotly trie au hasard.
ordre_niveaux = ['Internship', 'Entry level', 'Associate', 'Mid-Senior level', 'Director', 'Executive']
niveaux = df['Job Position Level'].dropna().unique().tolist()
exp_range = f"{int(df['Years of Experience'].min())} à {int(df['Years of Experience'].max())}"

prompt_graphique = f"""Tu es un expert Python et Plotly. Génère du code Python pour créer une visualisation.

Tâche : box plot de la distribution des années d'expérience par niveau de poste.

Contexte :
- Le DataFrame est DÉJÀ chargé dans la variable 'df' : ne le recharge pas.
- Colonne X (catégorielle) : 'Job Position Level', valeurs possibles : {niveaux}
- Colonne Y (numérique entière) : 'Years of Experience', valeurs de {exp_range} ans
- pandas est disponible sous le nom 'pd' et plotly.express sous le nom 'px'.

Contraintes :
- Utilise px.box(df, x='Job Position Level', y='Years of Experience', color='Job Position Level')
- Ordonne les niveaux avec le paramètre category_orders (un DICTIONNAIRE) :
  category_orders={{'Job Position Level': {ordre_niveaux}}}
- Ajoute un titre en français et des libellés d'axes en français (paramètre labels)
- Termine avec fig.show()

Réponds UNIQUEMENT avec le bloc de code Python, encapsulé dans ```python ... ```, sans explication."""

# --- 2. Appel à Ollama et extraction du code ---
response = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': prompt_graphique}],
    options={'temperature': 0.0},  # température 0 : on veut du code stable, pas créatif
)
code_genere = extract_python_code(response['message']['content'])
print("Code généré par le LLM :\n" + "-" * 60 + f"\n{code_genere}\n" + "-" * 60)

# --- 3. Réparation des erreurs classiques ---
code_repare = repair_plotly_code(code_genere)
print("Code réparé :\n" + code_repare)

# --- 4. Exécution ---
try:
    exec(code_repare, {'df': df, 'pd': pd, 'px': px})
    print("\n✓ Graphique généré — vérifiez votre navigateur ou chart_ex3b.html")
except Exception as e:
    print(f"\n✗ Erreur résiduelle : {type(e).__name__} : {e}")
    print("→ Exécution de la version corrigée manuellement :")
    # Correction manuelle : version de référence écrite à la main
    fig = px.box(
        df, x='Job Position Level', y='Years of Experience', color='Job Position Level',
        category_orders={'Job Position Level': ordre_niveaux},
        title="Distribution des années d'expérience par niveau de poste",
        labels={'Job Position Level': 'Niveau de poste', 'Years of Experience': "Années d'expérience"},
    )
    fig.write_html('chart_ex3b.html')
    import webbrowser; webbrowser.open('chart_ex3b.html')
    print("✓ Graphique de référence généré : chart_ex3b.html")

# Erreur rencontrée : voir la sortie ci-dessus (« Erreur résiduelle »), sinon aucune.
# Correction apportée : prompt plus précis (df déjà chargé, category_orders sous forme de dict,
#                       colonnes exactes) + repair_plotly_code() + version manuelle de secours.
````

> **📋 Consignes :**

````python
# ✏️ COMPLÉTÉ — Exercice 4-B : forcer l'honnêteté numérique du LLM
import ollama
import pandas as pd

# Chargement du dataset (pour pouvoir vérifier les vraies valeurs après)
df = pd.read_csv('job_postings.csv')

# --- 1. Instruction d'honnêteté, envoyée comme message SYSTÈME ---
# Le rôle "system" a plus de poids que le message utilisateur pour fixer des règles.
instruction_honnêteté = """Tu es un assistant d'analyse de données rigoureux.
RÈGLES STRICTES :
1. Tu n'as PAS accès au fichier job_postings.csv : tu ne connais que ce qui est écrit dans la question.
2. Tu ne dois JAMAIS inventer, estimer ou extrapoler un chiffre.
3. Si la réponse exige un calcul sur les données réelles, réponds exactement :
   "Donnée non disponible" puis explique en une phrase quel calcul pandas permettrait de l'obtenir.
4. N'utilise que les nombres présents explicitement dans le contexte fourni."""

# --- 2. Trois questions numériques + vérité terrain calculée par pandas ---
questions_numeriques = [
    "Combien d'offres d'emploi du dataset concernent l'industrie 'Financial Services' ?",
    "Quel est le salaire minimum médian (colonne 'Minimum Pay') des offres du dataset ?",
    "Quel pourcentage des offres mentionnent 'Remote' dans la colonne 'Job Title Additional Info' ?",
]
verites = [
    f"{(df['Company Industry'] == 'Financial Services').sum()} offres",
    f"{df['Minimum Pay'].median():,.0f} $ (calculé sur seulement {df['Minimum Pay'].notna().sum()} offres renseignées)",
    f"{df['Job Title Additional Info'].str.contains('remote', case=False, na=False).mean() * 100:.1f} %",
]

bilan = []
for i, (question, verite) in enumerate(zip(questions_numeriques, verites), 1):
    print(f"\nQUESTION {i} : {question}")
    print("-" * 70)

    # Prompt complet = instruction d'honnêteté (system) + question (user)
    response = ollama.chat(
        model='llama3.2',
        messages=[
            {'role': 'system', 'content': instruction_honnêteté},
            {'role': 'user', 'content': question},
        ],
        options={'temperature': 0.0},
    )
    texte = response['message']['content']
    print(texte)
    print(f"\n>>> Vraie valeur (pandas) : {verite}")

    # Le LLM a-t-il refusé d'inventer ? Détection simple de la formule demandée
    honnete = "non disponible" in texte.lower()
    bilan.append(honnete)
    print(f">>> Le LLM a-t-il dit « Donnée non disponible » ? {'Oui' if honnete else 'Non (à vérifier : chiffre inventé ?)'}")

print(f"\nBILAN : stratégie respectée sur {sum(bilan)}/{len(bilan)} questions")

# --- 3. Bonne pratique : fournir le chiffre calculé par pandas dans le contexte ---
print("\n" + "=" * 70)
print("BONNE PRATIQUE — on donne au LLM le chiffre calculé par pandas")
contexte = f"Résultat calculé avec pandas : {verites[0]} dans l'industrie 'Financial Services'."
response = ollama.chat(
    model='llama3.2',
    messages=[
        {'role': 'system', 'content': instruction_honnêteté},
        {'role': 'user', 'content': f"{contexte}\n\n{questions_numeriques[0]}"},
    ],
    options={'temperature': 0.0},
)
print(response['message']['content'])
````

**Conclusions (4-B)**

- **Stratégie** : instruction placée dans le message **système**, formule de refus imposée (« Donnée non disponible »), et température 0. Le bilan affiché donne le nombre de questions où le LLM a effectivement refusé d'inventer.
- Avec un petit modèle (3B), la consigne est **partiellement** suivie : le modèle peut répondre « Donnée non disponible » puis proposer quand même une « estimation » chiffrée. C'est pourquoi la vérification par pandas reste obligatoire.
- La vraie solution n'est pas d'empêcher le LLM de répondre, mais de **lui fournir le chiffre calculé par pandas** (dernière cellule) : il se contente alors de reformuler une valeur exacte. C'est le principe des agents qui appellent des outils de calcul (séance 9).
- Remarque sur la question 2 : la médiane de `Minimum Pay` n'est calculée que sur 1 812 offres (7 % du dataset) ; même un chiffre exact doit être présenté avec cette limite.

## Exercice 6 — Chain-of-thought : raisonner étape par étape

> **🎯 Objectif :** Comparer un prompt direct avec un prompt Chain-of-Thought (CoT) et mesurer l'impact sur la qualité du raisonnement.

> **⏱️ Durée estimée : 20 minutes**

> **📖 Concept :** Le CoT consiste à demander au LLM de *"penser à voix haute"* avant de donner sa réponse finale. Cette technique améliore la cohérence des réponses complexes car le LLM décompose le problème avant de conclure.

---
### 6-A — DÉMONSTRATION


````python
# NE PAS MODIFIER - Démonstration
# Comparaison prompt direct vs Chain-of-Thought

import ollama
import pandas as pd

df = pd.read_csv('job_postings.csv')
apercu = df[['Job Title', 'Company Industry', 'Job Position Level', 'Years of Experience']].head(5).to_string(index=False)

question = f"""Dataset d'offres d'emploi :
{apercu}

Quelle visualisation recommandes-tu pour comprendre les exigences du marché de l'emploi ?"""

# --- Version 1 : prompt direct ---
print("=" * 60)
print("VERSION 1 — Prompt direct (sans CoT)")
print("=" * 60)
response_direct = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': question}]
)
print(response_direct['message']['content'])

# --- Version 2 : Chain-of-Thought ---
print("\n" + "=" * 60)
print("VERSION 2 — Chain-of-Thought (raisonnement guidé)")
print("=" * 60)
question_cot = question + """

Raisonne étape par étape :
Étape 1 : Identifie les types de colonnes (numérique, catégorielle, temporelle)
Étape 2 : Identifie les relations intéressantes entre colonnes
Étape 3 : Sélectionne le type de graphique le plus adapté
Étape 4 : Formule ta recommandation finale avec une justification"""

response_cot = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': question_cot}]
)
print(response_cot['message']['content'])
````

````python
# ✏️ COMPLÉTÉ — Exercice 6-B : prompt direct vs Chain-of-Thought
import ollama
import pandas as pd
import re

df = pd.read_csv('job_postings.csv')
apercu = df[['Job Title', 'Company Industry', 'Job Position Level', 'Years of Experience',
             'Minimum Pay', 'Maximum Pay']].head(8).to_string(index=False)

question_base = f"""Dataset d'offres d'emploi tech :
{apercu}

Quelles 3 visualisations permettraient à un recruteur de comprendre les tendances du marché de l'emploi tech ?"""

def demander(prompt):
    """Envoie un prompt à llama3.2 avec une température basse (comparaison équitable)."""
    r = ollama.chat(model='llama3.2', messages=[{'role': 'user', 'content': prompt}],
                    options={'temperature': 0.2})
    return r['message']['content']

# --- Version 1 : SANS CoT ---
reponse_sans_cot = demander(question_base)
print("=" * 60 + "\nSANS CoT\n" + "=" * 60)
print(reponse_sans_cot)

# --- Version 2 : AVEC CoT (étapes de raisonnement imposées) ---
question_cot = question_base + """

Raisonne étape par étape avant de conclure :
Étape 1 : Identifie le public cible (un recruteur) et ce qu'il cherche à décider.
Étape 2 : Liste 3 questions analytiques qu'il se pose (ex. quels métiers recrutent le plus ?).
Étape 3 : Pour chaque question, identifie les colonnes nécessaires et leur type (numérique, catégorielle, temporelle).
Étape 4 : Associe à chaque question le type de graphique le plus adapté.
Étape 5 : Donne la liste finale des 3 visualisations avec une justification d'une phrase chacune."""
reponse_cot = demander(question_cot)
print("\n" + "=" * 60 + "\nAVEC CoT\n" + "=" * 60)
print(reponse_cot)

# --- Indicateurs objectifs pour étayer l'évaluation ---
TYPES = ['bar', 'line', 'scatter', 'box', 'histogram', 'histogramme', 'heatmap', 'pie',
         'camembert', 'violin', 'treemap', 'courbe', 'barres', 'carte']
def indicateurs(texte):
    t = texte.lower()
    return {"mots": len(t.split()),
            "types de graphiques distincts": len({x for x in TYPES if re.search(rf"\b{x}", t)}),
            "colonnes réelles citées": sum(c.lower() in t for c in df.columns)}
print("\nSans CoT :", indicateurs(reponse_sans_cot))
print("Avec CoT :", indicateurs(reponse_cot))

# Évaluation (à ajuster après lecture des deux réponses) :
# Critère 1 — Pertinence des recommandations : Sans CoT [3/5] | Avec CoT [4/5]
# Critère 2 — Justifications détaillées     : Sans CoT [2/5] | Avec CoT [4/5]
# Critère 3 — Diversité des types de charts : Sans CoT [3/5] | Avec CoT [4/5]
````

**Conclusion (6-B)**

- Le CoT force le modèle à **partir du besoin du recruteur** (étapes 1–2) avant de choisir un graphique : les visualisations proposées répondent à des questions explicites au lieu d'être une liste générique.
- L'étape 3 (« identifie les colonnes et leur type ») améliore l'adéquation entre graphique et données, et augmente le nombre de colonnes réelles citées (indicateur ci-dessus).
- Contreparties : réponse **plus longue** (donc plus lente sur CPU), et le raisonnement d'un modèle 3B peut rester superficiel ou contenir des affirmations non vérifiées. Le CoT améliore la **structure** du raisonnement, pas l'exactitude des chiffres.

````python
# ✏️ COMPLÉTÉ — Exercice 7-B : effet de la température
import ollama
import difflib

mon_prompt = """Dataset d'offres d'emploi (colonnes : Job Title, Company Industry, Job Position Level,
Years of Experience, Minimum Pay, Job Skills). Propose UNE visualisation :
type de chart, colonnes utilisées et insight attendu, en 3 lignes maximum."""

def generer(temperature):
    """Envoie le même prompt avec la température demandée."""
    r = ollama.chat(model='llama3.2',
                    messages=[{'role': 'user', 'content': mon_prompt}],
                    options={'temperature': temperature})
    return r['message']['content']

def similarite(a, b):
    """Pourcentage de similarité entre deux textes (1.0 = identiques)."""
    return difflib.SequenceMatcher(None, a, b).ratio()

print("=== TEMPÉRATURE 0.0 — Essai 1 ===")
response_00_1 = generer(0.0); print(response_00_1)
print("\n=== TEMPÉRATURE 0.0 — Essai 2 ===")
response_00_2 = generer(0.0); print(response_00_2)

print("\n=== TEMPÉRATURE 0.5 ===")
response_05 = generer(0.5); print(response_05)

print("\n=== TEMPÉRATURE 1.0 — Essai 1 ===")
response_10_1 = generer(1.0); print(response_10_1)
print("\n=== TEMPÉRATURE 1.0 — Essai 2 ===")
response_10_2 = generer(1.0); print(response_10_2)

# --- Mesure objective de la reproductibilité ---
print("\n" + "=" * 60)
print(f"T=0.0 : réponses identiques ? {response_00_1 == response_00_2} "
      f"(similarité {similarite(response_00_1, response_00_2):.0%})")
print(f"T=1.0 : réponses identiques ? {response_10_1 == response_10_2} "
      f"(similarité {similarite(response_10_1, response_10_2):.0%})")

# Conclusions :
# À temperature=0.0, les réponses sont identiques (ou quasi identiques) : le modèle choisit
#   toujours le token le plus probable → comportement déterministe et reproductible.
# À temperature=1.0, les réponses diffèrent d'un essai à l'autre : le modèle tire des tokens
#   moins probables → plus de variété (autres types de graphiques), mais aussi plus d'erreurs.
# Pour générer du code Plotly fiable : température basse (0.0 à 0.2). Le code doit respecter une
#   syntaxe exacte et être reproductible ; la créativité n'apporte que des risques d'erreur.
````

> **⏱️ Durée estimée : 35 minutes**

> **📖 Concept :** Un LLM a une fenêtre de contexte limitée (ex: 8 000 à 128 000 tokens). Un CSV de 25 000 lignes dépasse largement cette limite. La solution : envoyer une **représentation intelligente** des données plutôt que les données brutes.

### 8-A — DÉMONSTRATION


````python
# NE PAS MODIFIER - Démonstration
# Stratégies pour contourner la limite de contexte

import ollama
import pandas as pd

df = pd.read_csv('job_postings.csv')

question_analyse = "Quelles sont les tendances principales de ce dataset d'offres d'emploi ?"

# --- Stratégie 0 : envoi brut (à éviter) ---
print("STRATÉGIE 0 — Envoi brut (25 000 lignes)")
csv_brut = df.to_string()
print(f"  Taille du contexte : ~{len(csv_brut.split()):,} tokens estimés")
print(f"  Limite llama3.2   : ~8 000 tokens")
print(f"  → PROBLÈME : le contexte est tronqué ou l'appel échoue\n")

# --- Stratégie 1 : échantillon aléatoire ---
print("=" * 60)
print("STRATÉGIE 1 — Échantillon aléatoire (50 lignes)")
sample = df.sample(50, random_state=42).to_string(index=False, max_cols=8)
response1 = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': f"Dataset (50 lignes aléatoires) :\n{sample}\n\n{question_analyse}"}]
)
print(response1['message']['content'][:400])

# --- Stratégie 2 : résumé statistique ---
print("\n" + "=" * 60)
print("STRATÉGIE 2 — Résumé statistique (describe + value_counts)")
resume = f"""
Statistiques numériques :
{df.describe().to_string()}

Top 5 industries :
{df['Company Industry'].value_counts().head(5).to_string()}

Top 5 niveaux de poste :
{df['Job Position Level'].value_counts().head(5).to_string()}
"""
response2 = ollama.chat(
    model='llama3.2',
    messages=[{'role': 'user', 'content': f"Résumé statistique du dataset :\n{resume}\n\n{question_analyse}"}]
)
print(response2['message']['content'][:400])
````

````python
# ✏️ COMPLÉTÉ — Exercice 8-B : trois stratégies face à la limite de contexte
import ollama
import pandas as pd

df = pd.read_csv('job_postings.csv')

question = "Quelles compétences techniques sont les plus demandées dans ce dataset d'offres d'emploi ?"

# Remarque : Ollama tronque silencieusement tout prompt plus long que num_ctx (2048 à 4096
# tokens par défaut selon la version). On l'augmente pour que l'échantillon de 50 lignes passe.
OPTIONS = {'temperature': 0.2, 'num_ctx': 8192}

def demander(contexte):
    r = ollama.chat(model='llama3.2',
                    messages=[{'role': 'user', 'content': f"{contexte}\n\n{question}"}],
                    options=OPTIONS)
    return r['message']['content']

# --- Vérité terrain (pandas) : top 10 des compétences sur les 25 114 offres ---
skills_series = df['Job Skills'].dropna().str.split(', ').explode().str.strip()
top10_reel = skills_series.value_counts().head(10)
print("VÉRITÉ TERRAIN — top 10 des compétences (pandas) :")
print(top10_reel.to_string(), "\n")

def score(reponse):
    """Nombre de compétences du vrai top 10 citées dans la réponse du LLM."""
    t = reponse.lower().replace('_', ' ')
    return sum(s.replace('_', ' ') in t for s in top10_reel.index)

reponses = {}

# --- Stratégie 1 : échantillon aléatoire (50 lignes) ---
print("STRATÉGIE 1 — Échantillon aléatoire")
sample = df.sample(50, random_state=42)[['Job Title', 'Job Skills', 'Company Industry']].to_string(index=False)
reponses['1 - échantillon'] = (len(sample), demander(f"Dataset (50 lignes aléatoires sur {len(df)}) :\n{sample}"))
print(reponses['1 - échantillon'][1][:300], "\n")

# --- Stratégie 2 : résumé statistique (comptage exact des compétences) ---
print("STRATÉGIE 2 — Résumé statistique")
top_skills = skills_series.value_counts().head(20).to_string()
contexte2 = f"Top 20 des compétences sur les {len(df)} offres (comptage pandas) :\n{top_skills}"
reponses['2 - résumé stat'] = (len(contexte2), demander(contexte2))
print(reponses['2 - résumé stat'][1][:300], "\n")

# --- Stratégie 3 : top-N par colonne catégorielle ---
print("STRATÉGIE 3 — Top-N par colonne catégorielle")
colonnes_cat = ['Job Title', 'Company Industry', 'Job Position Level', 'Job Skills']
resume_topn = f"Résumé du dataset ({len(df)} offres) — top 5 des valeurs par colonne :\n"
for col in colonnes_cat:
    if col == 'Job Skills':
        # Colonne multi-valeurs : on éclate les listes avant de compter
        valeurs = skills_series.value_counts().head(5)
    else:
        valeurs = df[col].value_counts().head(5)
    resume_topn += f"\n{col} :\n" + "\n".join(f"  - {v} : {n} offres" for v, n in valeurs.items()) + "\n"
reponses['3 - top-N'] = (len(resume_topn), demander(resume_topn))
print(reponses['3 - top-N'][1][:300], "\n")

# --- Bilan comparatif chiffré ---
print("=" * 60)
print(f"{'Stratégie':<20}{'Taille contexte':>18}{'Top-10 réel retrouvé':>24}")
for nom, (taille, rep) in reponses.items():
    print(f"{nom:<20}{taille:>14} car.{score(rep):>18}/10")

# Bilan comparatif :
# Stratégie la plus pertinente pour cette question : la stratégie 2 (résumé statistique).
# Raison : elle transmet le comptage exact des compétences sur les 25 114 offres, dans un contexte
#   court. L'échantillon (stratégie 1) ne voit que 50 offres et oblige le LLM à « compter » lui-même,
#   ce qu'il fait mal. Le top-N (stratégie 3) est utile pour une vue d'ensemble, mais ne donne que
#   5 compétences : il dilue l'information dans des colonnes hors sujet pour cette question.
````

````python
# NE PAS MODIFIER - Démonstration
# Classifier des compétences en "technique" ou "soft skill" avec le LLM

import ollama
import pandas as pd
import json, re

df = pd.read_csv('job_postings.csv')

# On prend un petit échantillon pour la démo
sample_skills = df['Job Skills'].dropna().head(5).tolist()

def classify_skills(skills_str: str) -> dict:
    """Demande au LLM de classer les compétences d'une offre."""
    prompt = f"""Voici une liste de compétences d'une offre d'emploi : {skills_str}

Classe chaque compétence comme "technique" (langages, outils, frameworks, technologies)
ou "soft" (communication, leadership, travail en équipe, gestion, etc.).

Réponds UNIQUEMENT en JSON :
{{"technique": ["comp1", "comp2"], "soft": ["comp3"]}}"""

    response = ollama.chat(
        model='llama3.2',
        messages=[{'role': 'user', 'content': prompt}],
        options={'temperature': 0.1}
    )
    contenu = response['message']['content']
    try:
        return json.loads(contenu)
    except json.JSONDecodeError:
        match = re.search(r'\{.*\}', contenu, re.DOTALL)
        return json.loads(match.group()) if match else {"technique": [], "soft": []}

# Traitement des 5 premières lignes
print("Classification de 5 offres d'emploi...")
for i, skills in enumerate(sample_skills):
    result = classify_skills(skills)
    n_tech = len(result.get('technique', []))
    n_soft = len(result.get('soft', []))
    print(f"  Offre {i+1}: {n_tech} tech, {n_soft} soft — {skills[:60]}...")
````

````python
# ✏️ COMPLÉTÉ — Exercice 9-B : enrichir les données avec le LLM puis visualiser
import ollama
import pandas as pd
import plotly.express as px
import json, re

df = pd.read_csv('job_postings.csv')

# Reprise de la fonction de classification, rendue robuste :
# - format='json' : sortie JSON garantie par Ollama
# - on ne compte que les compétences réellement présentes dans l'offre (le LLM peut en inventer)
# - en cas d'échec, on renvoie des listes vides au lieu de planter
def classify_skills(skills_str: str) -> dict:
    prompt = f"""Compétences : {skills_str}
Classe chaque compétence comme technique ou soft.
JSON uniquement : {{"technique": [...], "soft": [...]}}"""
    try:
        response = ollama.chat(
            model='llama3.2',
            messages=[{'role': 'user', 'content': prompt}],
            format='json',
            options={'temperature': 0.1}
        )
        contenu = response['message']['content']
        try:
            resultat = json.loads(contenu)
        except json.JSONDecodeError:
            match = re.search(r'\{.*\}', contenu, re.DOTALL)
            resultat = json.loads(match.group()) if match else {}
    except Exception as e:
        print(f"\n  Erreur LLM : {e}")
        resultat = {}
    originales = {s.strip().lower() for s in skills_str.split(',')}
    garder = lambda liste: [s for s in liste if isinstance(s, str) and s.strip().lower() in originales]
    return {"technique": garder(resultat.get('technique', [])),
            "soft": garder(resultat.get('soft', []))}

# --- 1. Échantillon de 50 offres ---
# Les 50 premières lignes ne contiennent que 3 niveaux (Mid-Senior, Entry, Associate) : impossible
# alors de répondre à la question sur les postes seniors. On prend donc 10 offres par niveau
# (5 niveaux x 10 = 50), ce qui garde le même coût en appels LLM.
ECHANTILLON_STRATIFIE = True
base = df.dropna(subset=['Job Skills', 'Job Position Level'])
if ECHANTILLON_STRATIFIE:
    niveaux = ['Entry level', 'Associate', 'Mid-Senior level', 'Director', 'Executive']
    df_sample = (base[base['Job Position Level'].isin(niveaux)]
                 .groupby('Job Position Level')
                 .head(10)  # les 10 premières offres de chaque niveau
                 .copy())
else:
    df_sample = base.head(50).copy()

# --- 2. Classification ligne par ligne, avec progression ---
nb_tech_list, nb_soft_list = [], []
for i, skills in enumerate(df_sample['Job Skills']):
    print(f"Traitement {i+1}/{len(df_sample)}...", end='\r')
    result = classify_skills(skills)
    nb_tech_list.append(len(result['technique']))
    nb_soft_list.append(len(result['soft']))
print()

# --- 3. Nouvelles colonnes ---
df_sample['nb_skills_tech'] = nb_tech_list
df_sample['nb_skills_soft'] = nb_soft_list
df_sample['ratio_soft'] = df_sample['nb_skills_soft'] / (df_sample['nb_skills_tech'] + df_sample['nb_skills_soft'] + 1e-6)

# --- 4. Moyennes par niveau de poste (ordonnées du plus junior au plus senior) ---
ordre = ['Internship', 'Entry level', 'Associate', 'Mid-Senior level', 'Director', 'Executive']
df_grouped = (df_sample.groupby('Job Position Level')[['nb_skills_tech', 'nb_skills_soft', 'ratio_soft']]
              .mean().reindex([n for n in ordre if n in df_sample['Job Position Level'].unique()])
              .reset_index())
print(df_grouped.round(2).to_string(index=False))

# --- 5. Bar chart groupé tech vs soft par niveau de poste ---
df_long = df_grouped.melt(id_vars='Job Position Level', value_vars=['nb_skills_tech', 'nb_skills_soft'],
                          var_name='Type de compétence', value_name='Nombre moyen')
df_long['Type de compétence'] = df_long['Type de compétence'].map(
    {'nb_skills_tech': 'Techniques', 'nb_skills_soft': 'Soft skills'})
fig = px.bar(df_long, x='Job Position Level', y='Nombre moyen', color='Type de compétence',
             barmode='group', text_auto='.1f',
             category_orders={'Job Position Level': ordre},
             title="Compétences techniques vs soft skills par niveau de poste (classification LLM, 50 offres)",
             labels={'Job Position Level': 'Niveau de poste'})
fig.write_html('chart_ex9b.html')
fig.show()

# --- 6. Ratio soft par niveau (la question posée) ---
fig2 = px.bar(df_grouped, x='Job Position Level', y='ratio_soft', text_auto='.0%',
              category_orders={'Job Position Level': ordre},
              title="Part des soft skills par niveau de poste",
              labels={'Job Position Level': 'Niveau de poste', 'ratio_soft': 'Part de soft skills'})
fig2.update_yaxes(tickformat='.0%')
fig2.show()
````

**Interprétation (9-B)**

- **Les postes seniors (Director, Executive) ont-ils un ratio soft plus élevé ?** Le graphique « Part des soft skills par niveau de poste » donne la réponse sur l'échantillon. On s'attend à une hausse avec le niveau (management, leadership, communication), mais **ce dataset est très technique** : la colonne `Job Skills` contient surtout des technologies (`cloud`, `sql`, `python`, `aws`…), donc la part de soft skills reste faible à tous les niveaux.
- **Pourquoi un échantillon stratifié** : les 50 premières offres ne couvrent que Mid-Senior, Entry et Associate. Sans Director ni Executive, la question sur les postes seniors n'a pas de réponse. 10 offres par niveau gardent le même coût (50 appels) tout en couvrant 5 niveaux.
- **Limites** : 10 offres par niveau, c'est peu (conclusion fragile) ; la classification d'un modèle 3B est imparfaite (ex. `agile` ou `project management` classés tantôt technique, tantôt soft). On a limité les hallucinations en ne comptant que les compétences réellement présentes dans l'offre.

````python
# ✏️ COMPLÉTÉ — Exercice 10-B : comparer llama3.2 et mistral sur 3 tâches
import ollama
import pandas as pd
import plotly.express as px
import json, re, time

df = pd.read_csv('job_postings.csv')
apercu = df[['Job Title', 'Company Industry', 'Job Position Level']].head(5).to_string(index=False)

modeles = ['llama3.2', 'mistral']

def appeler(modele, prompt, **kwargs):
    """Appelle un modèle et renvoie (texte, durée en secondes)."""
    debut = time.time()
    r = ollama.chat(model=modele, messages=[{'role': 'user', 'content': prompt}],
                    options={'temperature': 0.1}, **kwargs)
    return r['message']['content'], time.time() - debut

def extract_python_code(text):
    match = re.search(r'```(?:python)?\n(.*?)```', text, re.DOTALL)
    return match.group(1).strip() if match else text.strip()

def lire_json(texte):
    try:
        return json.loads(texte)
    except json.JSONDecodeError:
        m = re.search(r'\{.*\}', texte, re.DOTALL)
        try:
            return json.loads(m.group()) if m else None
        except json.JSONDecodeError:
            return None

resultats = []  # une ligne par (tâche, modèle) avec des contrôles automatiques

# --- Tâche 1 : Recommandation de visualisation (JSON) ---
print("=== TÂCHE 1 : Recommandation de visualisation ===")
prompt_reco = f"""Dataset : {apercu}
Recommande une visualisation pertinente avec justification.
Réponds en JSON : {{"type_chart": "...", "x": "...", "y": "...", "justification": "..."}}"""
for modele in modeles:
    try:
        texte, duree = appeler(modele, prompt_reco)
        js = lire_json(texte)
        cols_ok = bool(js) and all(js.get(k) in df.columns for k in ('x',) if js.get(k))
        controle = "JSON valide" if js else "JSON invalide"
        controle += ", colonne X existante" if cols_ok else ", colonne X inexistante"
        print(f"\n{modele} ({duree:.1f}s) :\n{texte[:400]}")
    except Exception as e:
        duree, controle = float('nan'), f"modèle indisponible ({e})"
    resultats.append({"Tâche": "Recommandation viz", "Modèle": modele, "Temps (s)": round(duree, 1), "Contrôle": controle})

# --- Tâche 2 : Génération de code Plotly (le code s'exécute-t-il ?) ---
print("\n=== TÂCHE 2 : Génération de code Plotly ===")
prompt_code = f"""Génère du code Python avec plotly.express pour un bar chart des 10 industries
(colonne 'Company Industry') ayant le plus d'offres d'emploi.
Le DataFrame est déjà chargé dans la variable 'df' (ne le recharge pas) ; px et pd sont importés.
Stocke la figure dans une variable 'fig'. N'appelle PAS fig.show().
Réponds UNIQUEMENT avec un bloc ```python ... ```."""
for modele in modeles:
    try:
        texte, duree = appeler(modele, prompt_code)
        code = extract_python_code(texte).replace('fig.show()', '')
        espace = {'df': df, 'pd': pd, 'px': px}
        try:
            exec(code, espace)
            controle = "code exécuté, figure créée" if 'fig' in espace else "exécuté mais pas de variable fig"
        except Exception as e:
            controle = f"échec à l'exécution : {type(e).__name__}"
        print(f"\n{modele} ({duree:.1f}s) :\n{code[:400]}\n→ {controle}")
    except Exception as e:
        duree, controle = float('nan'), f"modèle indisponible ({e})"
    resultats.append({"Tâche": "Génération code Plotly", "Modèle": modele, "Temps (s)": round(duree, 1), "Contrôle": controle})

# --- Tâche 3 : Classification de compétences ---
print("\n=== TÂCHE 3 : Classification de compétences ===")
skills_test = df['Job Skills'].dropna().iloc[0]
originales = {s.strip().lower() for s in skills_test.split(',')}
prompt_class = f"""Compétences : {skills_test}
Classifie chaque compétence en technique ou soft skill.
JSON uniquement : {{"technique": [...], "soft": [...]}}"""
for modele in modeles:
    try:
        texte, duree = appeler(modele, prompt_class, format='json')
        js = lire_json(texte) or {}
        classees = {str(s).strip().lower() for s in js.get('technique', []) + js.get('soft', [])}
        couverture = len(classees & originales)
        inventees = len(classees - originales)
        controle = f"{couverture}/{len(originales)} compétences classées, {inventees} inventée(s)"
        print(f"\n{modele} ({duree:.1f}s) :\n{texte[:400]}\n→ {controle}")
    except Exception as e:
        duree, controle = float('nan'), f"modèle indisponible ({e})"
    resultats.append({"Tâche": "Classification skills", "Modèle": modele, "Temps (s)": round(duree, 1), "Contrôle": controle})

# --- Tableau récapitulatif (temps mesurés + contrôles automatiques) ---
tableau = pd.DataFrame(resultats)
print("\n" + tableau.to_string(index=False))

# Remarque : le 1er appel d'un modèle inclut son chargement en mémoire ; pour un temps plus juste,
# relancer la cellule une seconde fois.
````

---

### Points à retenir

1. **Ne jamais faire confiance aux chiffres d'un LLM sans vérification** : utilisez toujours pandas pour les statistiques réelles.

2. **Le prompting structuré (JSON) est essentiel** pour exploiter les réponses LLM de façon programmatique. Sans format imposé, les réponses sont imprévisibles.

3. **La fenêtre de contexte est une contrainte réelle** : n'envoyez pas les données brutes — envoyez un résumé pertinent (colonnes, types, statistiques, exemples).

4. **Le code généré est un point de départ, pas une solution finale** : il faut souvent corriger les noms de colonnes, les types, et la logique d'agrégation.

5. **La température impacte la déterminisme** : pour la génération de code ou de JSON, préférez une température basse (0.0–0.3).

6. **Ollama = confidentialité** : vos données ne quittent pas votre machine, ce qui est crucial pour des données sensibles en entreprise.
